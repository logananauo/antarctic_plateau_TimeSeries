## **Antarctic Plateau Salt Relative Humidity and Temperature Dataset (APSRHT) - Cleaning**

Dataset: Davila, A; Parro, V. Antarctic Plateau-Salt Relative Humidity and Temperature. Astrobiology Habitable Environment Database. Retrieved Date: Jun 23, 2026. https://doi.org/10.48667/sy0j-yc84

"First column is the measurement number. Second column is the date/time of each measurement. Third column is the temperature in degrees Celssius. Fourth column is the RH as measured by the sensor. Fifth column is the RH with respect to ice"

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('Antarctic_Plateau_Salt data.csv', encoding='latin1')
df.columns = df.iloc[0]
df = df[1:].reset_index(drop=True)
df
```

```python
### rename columns
df = df.rename(columns={'#': 'measurement_number', 'Date Time': 'date_time', 'T (¼C)': 'temp_C', 'RH (%)': 'relative_hum_NaCl', 'RH (ice) (%)': 'relative_hum_Ice'})
df.head()
```

```python
### convert to datetime
df['date_time'] = pd.to_datetime(df['date_time'], format='%m/%d/%y %H:%M')

### sort and set index
df = df.sort_values('date_time')
df = df.set_index('date_time')

### set 15-minute frequency and fill missing logs
#### ARIMA will fail if a 15-minute slot is missing from the raw data
df = df.asfreq('15min')

### fill missing measurement values caused by missing slots
df = df.ffill() 
df.head()
```

```python
### write to .csv
df.to_csv('apsrht_datetime.csv', index=True, encoding='utf-8')
```
