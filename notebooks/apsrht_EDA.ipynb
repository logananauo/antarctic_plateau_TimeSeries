## **Antarctic Plateau Salt Relative Humidity and Temperature Dataset - Exploratory Data Analysis**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid', palette='deep')
```

```python
### load preprocessed data
data = pd.read_csv('apsrht_datetime.csv', index_col='date_time', parse_dates=True)

## reapply 15-min frequency 
data = data.asfreq('15min')
data.head()
```

```python
### null values
data.isna().sum()
```

```python
### sum stats
data.describe()
```

```python
### info
data.info()
```

#### **A Few Key Plots:**

```python
plt.figure(figsize=(14, 6))

sns.lineplot(data=data, 
             x='temp_C', 
             y='relative_hum_Ice', 
             color='#2b5c8f')

plt.title('Ice Relative Humidity  vs.  Temperature', 
          fontsize=16, 
          fontweight='bold', 
          pad=15)

plt.xlabel('Temperature $^\circ$C', fontsize=12, labelpad=10)
plt.ylabel('Relative Humidity (%)', fontsize=12, labelpad=10)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 6))
sns.lineplot(data=data, 
             x='measurement_number', 
             y='temp_C', 
             color='#2b5c8f')

plt.title('Temperature  vs.  Measurements', 
          fontsize=16, 
          fontweight='bold', 
          pad=15)

plt.xlabel('measurement number', fontsize=12, labelpad=10)
plt.ylabel('Temperature $^\circ$C', fontsize=12, labelpad=10)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 6))

sns.lineplot(data=data, 
             x='measurement_number', 
             y='relative_hum_Ice', 
             color='#2b5c8f')

plt.title('Ice Relative Humidity  vs.  Measurements', 
          fontsize=16, 
          fontweight='bold', 
          pad=15)

plt.xlabel('measurement number', fontsize=12, labelpad=10)
plt.ylabel('Relative Humidity (%)', fontsize=12, labelpad=10)
plt.tight_layout()
plt.show()
```
